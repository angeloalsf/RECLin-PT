**Execução exploratória pré-registrada (22/11/2026).** Repete a execução sem pesos de classe (`pair_aware__a3ac671a77e2__t2`, `class_weight="none"`, 3 épocas) mudando **um único parâmetro**: `epochs` 3 → **5**. Todo o resto é idêntico: estratégia `pair_aware`, cabeça linear `[h_cls ; h_E1 ; h_E2] → Dropout → Linear(3d, 3)`, partições v2, revisão dos pesos, `max_gap=25`, `batch_size=64`, `lr=2e-5`, `warmup_ratio=0.1`, semente 42, seleção da época pelo **macro-F1 do DEV completo** (entre as épocas 1 a 5).

**Por quê.** Em todas as execuções de três rótulos a melhor época foi a última permitida (a 3ª): o treino pode não ter estabilizado.

**Atenção:** não é a execução de 3 épocas "continuada". O agendador da taxa de aprendizado decai até zero no fim do treino; com 5 épocas a curva inteira muda, e mesmo a época 3 deste treino é outro modelo.

**Regra de decisão (aprovada pelo autor antes do treino; `docs/protocolo/emenda-configuracao-final.md`, §8).** Nos dois modelos (3 e 5 épocas), t é escolhido pelo mesmo critério — maior precisão de `negation_of` entre os limiares da grade 0,05…0,95 que recuperam pelo menos 80% das negações anotadas. **A versão de 5 épocas substitui a de 3 só se a precisão dela, no seu ponto escolhido, for maior, com o IC95% por documento (bootstrap pareado, 10.000 reamostras, semente 42) excluindo o zero.** Caso contrário, fica a de 3 épocas. A seção 8.3 aplica a regra e imprime a decisão.

**Só DEV.** Este notebook **não lê nem avalia o TEST** e **não congela** nada. O congelamento é feito depois, pelo notebook da configuração final, com a versão escolhida.

## 7. Treinar a Pair-Aware BioBERTpt-all **sem pesos de classe, 5 épocas** (com retomada)

**7a** fixa os parâmetros e confere que as únicas diferenças para o protocolo são `class_weight` (balanced → none) e `epochs` (3 → 5). **7b** decide sozinha o que fazer **para esta configuração**:

* há execução **interrompida compatível** (mesma configuração, dados, pesos e identidade do código) → **retoma** de onde parou;
* já há execução **concluída** desta configuração → **não treina de novo** (nada é sobrescrito); para outra tentativa, ponha `NOVA_TENTATIVA = True`;
* senão → cria a execução exploratória (`pair_aware__<config_id>__t1`).

A decisão olha só as execuções **desta** configuração: a pasta oficial `pairaware_biobertpt_seed42` (com `class_weight=balanced`) tem outro `config_id` e nunca é retomada, reaproveitada nem sobrescrita aqui.

Se o runtime cair, rode as seções 1–6 e depois 7a e 7b de novo.

```python
# 7a — parâmetros (rode também antes das seções 8–11 numa sessão nova: elas localizam a execução por esta configuração)
PARAMETROS = {                 # os da execução sem pesos (a3ac671a77e2) com UMA mudança: epochs 3 → 5
    'max_gap': 25, 'ctx_chars': 128, 'max_length': 128, 'epochs': 5, 'batch_size': 64, 'lr': 2e-5,
    'weight_decay': 0.01, 'warmup_ratio': 0.1, 'max_grad_norm': 1.0,
    'class_weight': 'none',        # como na execução sem pesos de 3 épocas
    'head_dropout': None,          # None = dropout do encoder
}
MOTIVO = ('Experimento pré-registrado (22/11/2026): Pair-Aware linear BioBERTpt-all sem pesos de classe com 5 épocas '
          '(único parâmetro alterado em relação a pair_aware__a3ac671a77e2__t2); regra de decisão na emenda da configuração final, §8')
NOVA_TENTATIVA = False         # True só para forçar OUTRA tentativa desta mesma configuração

from reclin.experimento import persistencia, proveniencia

_ctx_teste = _ctx(parametros=PARAMETROS, motivo=MOTIVO)
_cfg_teste = operacoes.config_da_operacao(_ctx_teste.espec, PARAMETROS).como_dict()
_cfg_prot = _ctx_teste.espec.config_protocolo().como_dict()
DIFERENCAS = {k: (_cfg_prot.get(k), v) for k, v in _cfg_teste.items() if _cfg_prot.get(k) != v}
print('Diferenças em relação ao protocolo (protocolo → teste):', DIFERENCAS)
assert set(DIFERENCAS) == {'class_weight', 'epochs'} and DIFERENCAS['class_weight'] == ('balanced', 'none') \
    and DIFERENCAS['epochs'] == (3, 5), 'Este notebook só pode ter class_weight=none e epochs=5; confira PARAMETROS.'

def _execucoes_desta_config():
    """As execuções (local e Drive) DESTA configuração, da tentativa mais recente para a mais antiga.
    Usa a mesma chave do plano de treino (configuração + dados + pesos + identidade do código)."""
    ctx = _ctx(parametros=PARAMETROS, motivo=MOTIVO)
    chave = operacoes._chave(ctx, operacoes.config_da_operacao(ctx.espec, PARAMETROS))
    return plano.compativeis(chave, RAIZ_LOCAL, RAIZ_DRIVE)

for e in _execucoes_desta_config():
    print(f"  já existe: {e['nome']} ({e['onde']}, {e['estado']})")
```

```python
# 7b — treinar / retomar / não treinar de novo
_compat = _execucoes_desta_config()
_ativas = [e for e in _compat if e['estado'] in ('interrompida', 'iniciada')]
_concluidas = [e for e in _compat if e['estado'] == 'concluida']
ctx = _ctx(parametros=PARAMETROS, motivo=MOTIVO)

if _ativas and not NOVA_TENTATIVA:
    OPERACAO = operacoes.OP_RETOMAR
    print('Há execução interrompida desta configuração:', _ativas[0]['nome'], '→ RETOMAR')
elif _concluidas and not NOVA_TENTATIVA:
    OPERACAO = None
    print('Esta configuração já foi treinada e concluída:', _concluidas[0]['nome'],
          '→ nada é treinado de novo. Siga para a seção 8. (Outra tentativa: NOVA_TENTATIVA = True.)')
else:
    OPERACAO = operacoes.OP_NOVA
    print('Nova execução EXPLORATÓRIA (class_weight=none, 5 épocas).')

if OPERACAO is not None:
    print('\nOperação:', OPERACAO, '| natureza: EXPLORATÓRIA (fora da fase científica, não elegível ao TEST)')
    QUADRO = apresentacao.quadro_pre_treino(ctx, OPERACAO, sha_codigo=SHA_CODIGO, repositorio=f'angeloalsf/{REPO_NAME}')
    try:
        apresentacao.exigir_quadro(QUADRO, ctx, OPERACAO)
    except operacoes.ErroOperacao as erro:
        raise SystemExit(f'ERRO: {erro}') from None
    RESULTADO_TREINO = operacoes.executar(OPERACAO, ctx)
```

## 8. Resultados (métricas + matriz de confusão)

```python
def _selecionar_execucao(restaurar=True):
    compat = _execucoes_desta_config()
    if not compat:
        raise SystemExit('Nenhuma execução desta configuração (class_weight=none, 5 épocas): treine primeiro (seção 7).')
    escolhida = ([e for e in compat if e['estado'] == 'concluida'] or compat)[0]
    nome = escolhida['nome']
    if escolhida['estado'] != 'concluida':
        print(f"ATENÇÃO: {nome} está '{escolhida['estado']}' (não concluída); termine o treino (7b) antes de ler os resultados.")
    if restaurar and not (RAIZ_LOCAL / nome / proveniencia.ARQUIVO).is_file():
        print(f'{nome} só existe no Drive: restaurando para o disco local…')
        persistencia.restaurar(RAIZ_DRIVE, nome, RAIZ_LOCAL, progresso=print)
    return nome

EXECUCAO_SELECIONADA = _selecionar_execucao()     # as seções 10 e 11 enviam SÓ esta execução
CTX = _ctx(execucao=EXECUCAO_SELECIONADA, parametros=PARAMETROS)
_reg = proveniencia.ler(apresentacao.pasta_local_da_execucao(CTX))
assert (_reg['parametros_de_treinamento'] or {}).get('class_weight') == 'none', 'A execução selecionada não é a de class_weight=none!'
assert (_reg['parametros_de_treinamento'] or {}).get('epochs') == 5, 'A execução selecionada não é a de 5 épocas!'
_base = CTX.espec.config_protocolo().como_dict()
NATUREZA = 'PROTOCOLO' if all(_reg['parametros_de_treinamento'].get(k) == _base[k] for k in CTX.espec.permitidos) else 'EXPLORATÓRIA'
print(f"Execução selecionada: {EXECUCAO_SELECIONADA} | natureza: {NATUREZA} | estado: {_reg['estado']}")
operacoes.executar(operacoes.OP_DEV, CTX)
```

### 8.3 Regra de decisão: 5 épocas × 3 épocas (pré-registrada)

Para cada modelo, t é escolhido pelo critério da configuração final (`reclin.estrategias.pair_aware_limiar`: grade 0,05…0,95; maior precisão de `negation_of` entre os limiares com *recall* ≥ 80%; desempates: maior F1, maior macro-F1, menor limiar). Depois, a diferença de **precisão de `negation_of`** (5 épocas no seu t − 3 épocas no seu t) é estimada por **bootstrap pareado por documento** (10.000 reamostras, semente 42; o mesmo cálculo de `diagnosticos/comparar_precisao_recall.py`). **5 épocas só substitui 3 se o IC95% por documento estiver todo acima de zero.** A célula também mostra recall, F1 e macro-F1, só para informação: eles não entram na regra. Só DEV.

```python
import importlib.util
import numpy as np
from reclin import tarefa as _tarefa
from reclin.avaliacao import limiar as _busca, metricas as _met
from reclin.estrategias import pair_aware_limiar as _pal
from reclin.execucao import predicoes as _pred
from reclin.util.io import ler_jsonl

TRES_EPOCAS = 'pair_aware__a3ac671a77e2__t2'
_c3 = comparacao_dev.localizar(RAIZES, TRES_EPOCAS)
_c5 = comparacao_dev.localizar(RAIZES, EXECUCAO_SELECIONADA)
if _c3 is None or _c5 is None:
    raise SystemExit('Predições não encontradas: ' + str([n for n, c in ((TRES_EPOCAS, _c3), (EXECUCAO_SELECIONADA, _c5))
                                                         if c is None]) + ' — nada foi calculado.')
p3, p5 = _pred.ler(_c3), _pred.ler(_c5)
assert _pred.mesmo_conjunto(p3, p5), 'y_true/labels diferem: as execuções não são comparáveis'
_cands = _tarefa.candidatos(list(ler_jsonl(Path(REPO) / 'codigo' / 'dados' / 'particoes-v2' / 'dev.jsonl')))
assert _tarefa.y_true(_cands) == list(p3['y_true']), 'o DEV reconstruído não confere com o y_true das predições'
GRUPOS = [c['doc_id'] for c in _cands]

def _escolher(p):
    t, tabela = _busca.buscar(p['y_true'], lambda x: _pal.prever(p['probs'], x), criterio=_pal.criterio())
    return t, tabela, _pal.prever(p['probs'], t)

T3, TAB3, Y3 = _escolher(p3)
T5, TAB5, Y5 = _escolher(p5)
print(f"{'':22s}{'t':>6s}{'VP':>6s}{'FP':>6s}{'FN':>6s}{'P':>8s}{'R':>8s}{'F1':>8s}{'macro':>8s}")
for nome, t, y in (('3 épocas', T3, Y3), ('5 épocas', T5, Y5)):
    m = _met.avaliar(p3['y_true'], y); n = m['por_classe']['negation_of']
    print(f"{nome:22s}{t:6.2f}{n['tp']:6d}{n['fp']:6d}{n['fn']:6d}{n['precision']:8.4f}{n['recall']:8.4f}"
          f"{n['f1']:8.4f}{m['macro_f1']:8.4f}")
if T3 != 0.8:
    print('\nATENÇÃO: o t de 3 épocas não deu 0,80; pare e envie a saída (as predições não são as analisadas).')

_spec = importlib.util.spec_from_file_location('cpr', Path(REPO) / 'codigo' / 'diagnosticos' / 'comparar_precisao_recall.py')
_cpr = importlib.util.module_from_spec(_spec); _spec.loader.exec_module(_cpr)
RES = _cpr.comparar(p3['y_true'], Y5, Y3, GRUPOS)            # A = 5 épocas, B = 3 épocas
print('\nDiferença 5 − 3 épocas, cada uma no seu t (DEV; IC95% por bootstrap pareado, 10.000 reamostras, semente 42)')
for k in ('precision', 'recall', 'f1'):
    d = RES['por_documento'][k]; c = RES['por_candidato'][k]
    print(f"  {k:9s} {d['diferenca']:+.4f} | doc [{d['ic95'][0]:+.4f}; {d['ic95'][1]:+.4f}] "
          f"| cand [{c['ic95'][0]:+.4f}; {c['ic95'][1]:+.4f}]")
ic = RES['por_documento']['precision']['ic95']
DECISAO = '5 épocas' if ic[0] > 0 else '3 épocas'
print('\nREGRA (pré-registrada): 5 épocas só se o IC95% por documento da precisão estiver todo acima de zero.')
print(f'DECISÃO: {DECISAO}  (IC95% por documento da diferença de precisão: [{ic[0]:+.4f}; {ic[1]:+.4f}])')
print('Envie esta saída: a decisão é registrada na emenda e no TCC antes do congelamento.')
```
